# Langkah Latihan


```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label

# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```

No 1


Perbedaan supervised, unsupervised, dan reinforcement learning adalah
- Supervised learning merupakan pembelajaran yang diarahkan. Belajar mencari hubungan dari atribut penjelas dengan jawaban serta memiliki label. Terdiri dari klasifikasi yang memiliki keluaran berupa kategori, dan regresi memiliki keluaran berupa nilai numerik. contoh: filter spam email.
- Unsupervised learning merupakan pembelajaran yang tidak memiliki label yang bertujuan untuk mencari hubungan atau pola antara data yang satu dengan data yang lain sehingga dapat menghasilkan sebuah kelompok. Contoh: segmentasi.
- Reinforcement learning adalah pembelajaran yang dilakukan berdasarkan interaksi melalui pemberian reward (penghargaan) atau punishment (hukuman). Contoh: agen game.

No 2


Pembagian data latih dan data uji penting untuk memastikan apakah model dapat mengenali data yang belum pernah dilihat sebelumnya. Data latih digunakan untuk memberikan pengetahuan atau melatih model, sedangkan data uji digunakan untuk mengetahui apakah model sudah dapat bekerja dengan baik ketika diuji menggunakan data baru.

Alur kerja ML disebut iteratif karena dalam proses pembuatan ML diperlukan evaluasi secara berulang. Jika model memiliki akurasi yang buruk atau belum dapat bekerja dengan baik, maka perlu dilakukan perbaikan dan pengujian kembali. Proses ini dilakukan berulang sampai model dapat bekerja dengan baik.
